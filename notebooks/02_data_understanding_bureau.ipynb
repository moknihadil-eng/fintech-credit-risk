```python
# ============================================================
# DATA PREPARATION + FEATURE ENGINEERING : BUREAU
# ============================================================

import pandas as pd
import numpy as np


# ------------------------------------------------------------
# 1) Chargement de la table bureau
# ------------------------------------------------------------

bureau = pd.read_csv(
    "data/raw/bureau.csv"
)

print("Dimension initiale :", bureau.shape)



# ------------------------------------------------------------
# 2) Création des indicateurs de valeurs manquantes
# ------------------------------------------------------------

missing_cols = [
    "AMT_ANNUITY",
    "AMT_CREDIT_MAX_OVERDUE",
    "AMT_CREDIT_SUM_LIMIT",
    "AMT_CREDIT_SUM_DEBT"
]


for col in missing_cols:

    bureau[col + "_MISSING"] = (
        bureau[col].isnull()
        .astype(int)
    )


print("Indicateurs missing créés")



# ------------------------------------------------------------
# 3) Traitement des valeurs manquantes
# ------------------------------------------------------------

# Absence de dette ou de limite = 0
zero_cols = [
    "AMT_CREDIT_MAX_OVERDUE",
    "AMT_CREDIT_SUM_LIMIT",
    "AMT_CREDIT_SUM_DEBT",
    "AMT_ANNUITY"
]


for col in zero_cols:

    bureau[col] = bureau[col].fillna(0)



# Dates manquantes :
# remplacement par la médiane
date_cols = [
    "DAYS_CREDIT_ENDDATE",
    "DAYS_ENDDATE_FACT"
]


for col in date_cols:

    bureau[col] = bureau[col].fillna(
        bureau[col].median()
    )



# Montant crédit manquant
bureau["AMT_CREDIT_SUM"] = bureau["AMT_CREDIT_SUM"].fillna(
    bureau["AMT_CREDIT_SUM"].median()
)



print(
    "Valeurs manquantes restantes :",
    bureau.isnull().sum().sum()
)



# ------------------------------------------------------------
# 4) Création des variables métier
# ------------------------------------------------------------


# Crédit actif

bureau["IS_ACTIVE"] = (
    bureau["CREDIT_ACTIVE"]
    =="Active"
).astype(int)



# Crédit en retard

bureau["IS_OVERDUE"] = (
    bureau["AMT_CREDIT_SUM_OVERDUE"] > 0
).astype(int)



# Ancienneté absolue du crédit

bureau["CREDIT_AGE_DAYS"] = (
    bureau["DAYS_CREDIT"]
    .abs()
)



# Dette / crédit
bureau["DEBT_RATIO"] = np.where(
    bureau["AMT_CREDIT_SUM"] != 0,
    bureau["AMT_CREDIT_SUM_DEBT"] /
    bureau["AMT_CREDIT_SUM"],
    0
)



# ------------------------------------------------------------
# 5) Agrégation par client
# ------------------------------------------------------------

bureau_features = bureau.groupby(
    "SK_ID_CURR"
).agg(


    # Nombre total de crédits
    NB_TOTAL_CREDITS = (
        "SK_ID_BUREAU",
        "count"
    ),


    # Nombre crédits actifs
    NB_ACTIVE_CREDITS = (
        "IS_ACTIVE",
        "sum"
    ),


    # Nombre crédits avec retard
    NB_OVERDUE_CREDITS = (
        "IS_OVERDUE",
        "sum"
    ),


    # Montant total crédit
    TOTAL_CREDIT = (
        "AMT_CREDIT_SUM",
        "sum"
    ),


    # Dette totale
    TOTAL_DEBT = (
        "AMT_CREDIT_SUM_DEBT",
        "sum"
    ),


    # Limite totale disponible
    TOTAL_CREDIT_LIMIT = (
        "AMT_CREDIT_SUM_LIMIT",
        "sum"
    ),


    # Retard total
    TOTAL_OVERDUE_AMOUNT = (
        "AMT_CREDIT_SUM_OVERDUE",
        "sum"
    ),


    # Crédit moyen
    AVG_CREDIT_AMOUNT = (
        "AMT_CREDIT_SUM",
        "mean"
    ),


    # Ancienneté moyenne
    AVG_CREDIT_AGE = (
        "CREDIT_AGE_DAYS",
        "mean"
    ),


    # Ratio dette/crédit moyen
    AVG_DEBT_RATIO = (
        "DEBT_RATIO",
        "mean"
    )

).reset_index()



# ------------------------------------------------------------
# 6) Vérification finale
# ------------------------------------------------------------


print("\n========== TABLE FINALE ==========")

print(
    bureau_features.shape
)


display(
    bureau_features.head()
)



print(
    "\nValeurs manquantes :",
    bureau_features.isnull().sum().sum()
)



# ------------------------------------------------------------
# 7) Sauvegarde
# ------------------------------------------------------------

bureau_features.to_csv(
    "data/processed/bureau_features.csv",
    index=False
)


print(
    "\nFichier sauvegardé : bureau_features.csv"
)
```

# Transformation de la table bureau au niveau client

## Objectif

La table `bureau` contient l'historique des crédits précédemment obtenus par les clients auprès d'autres organismes financiers.

Contrairement à la table `application_train` où chaque ligne représente un client, la table `bureau` contient plusieurs lignes pour un même client, car un client peut avoir plusieurs crédits historiques.

L'objectif est donc de transformer cette table afin d'obtenir une seule ligne par client (`SK_ID_CURR`) contenant des indicateurs synthétiques sur son comportement financier.

---

# Structure initiale de la table bureau

Avant transformation, chaque ligne correspond à un crédit individuel.

Exemple :

| SK_ID_CURR | SK_ID_BUREAU | CREDIT_ACTIVE | AMT_CREDIT_SUM | AMT_CREDIT_SUM_DEBT | AMT_CREDIT_SUM_OVERDUE |
|------------|--------------|---------------|----------------|---------------------|------------------------|
| 100001 | 5000001 | Active | 100000 | 30000 | 0 |
| 100001 | 5000002 | Closed | 50000 | 0 | 0 |
| 100001 | 5000003 | Active | 200000 | 80000 | 5000 |
| 100002 | 5000004 | Closed | 70000 | 0 | 0 |

Dans cet exemple :

- Le client `100001` possède 3 crédits historiques.
- Le client `100002` possède 1 crédit historique.

Cette structure n'est pas directement adaptée à un modèle de Machine Learning, car la variable cible `TARGET` est définie au niveau du client.

---

# Principe de transformation

Une agrégation par client (`SK_ID_CURR`) est réalisée afin de regrouper toutes les informations historiques d'un même client.

Les informations détaillées des crédits individuels sont transformées en variables comportementales.

---

# Exemple après transformation

Après agrégation, chaque client possède une seule ligne :

| SK_ID_CURR | NB_TOTAL_CREDITS | NB_ACTIVE_CREDITS | TOTAL_CREDIT | TOTAL_DEBT | TOTAL_OVERDUE_AMOUNT |
|------------|------------------|-------------------|--------------|------------|----------------------|
| 100001 | 3 | 2 | 350000 | 110000 | 5000 |
| 100002 | 1 | 0 | 70000 | 0 | 0 |

---

# Création des variables comportementales

## Nombre total de crédits

Cette variable représente le nombre de crédits historiques associés au client.

Exemple :

Client `100001` possède :

- crédit 1
- crédit 2
- crédit 3

Donc :
NB_TOTAL_CREDITS = 3


Cette variable permet d'évaluer l'expérience financière du client.

---

## Nombre de crédits actifs

Cette variable correspond au nombre de crédits encore ouverts.

Exemple :

Pour le client `100001` :

- Crédit 5000001 : Active
- Crédit 5000002 : Closed
- Crédit 5000003 : Active

Donc :
NB_ACTIVE_CREDITS = 2


Un nombre élevé de crédits actifs peut indiquer une charge financière importante.

---

## Montant total emprunté

La somme des montants de crédits historiques est calculée :

\[
TOTAL\_CREDIT = \sum AMT\_CREDIT\_SUM
\]

Exemple :

\[
100000 + 50000 + 200000 = 350000
\]

Cette variable représente l'exposition financière totale du client.

---

## Dette totale restante

La dette encore à rembourser est calculée :

\[
TOTAL\_DEBT = \sum AMT\_CREDIT\_SUM\_DEBT
\]

Exemple :

\[
30000 + 0 + 80000 = 110000
\]

Cette variable mesure le niveau d'endettement actuel.

---

## Montant total des retards

Les montants en retard sont additionnés :

\[
TOTAL\_OVERDUE\_AMOUNT =
\sum AMT\_CREDIT\_SUM\_OVERDUE
\]

Cette variable représente les difficultés de remboursement passées.

---

# Justification de la méthode

Cette transformation permet de passer :

- d'une vision **crédit individuel** :


1 716 428 lignes


à une vision **client** :


305 811 clients


Les nouvelles variables créées représentent le comportement financier global du client.

Elles sont donc plus pertinentes pour prédire le risque de défaut (`TARGET`) dans le modèle de Machine Learning.

---

# Résultat attendu

La table finale `bureau_features` contiendra :

- une ligne par client ;
- des variables numériques exploitables par les modèles ;
- aucune duplication de client ;
- des informations historiques synthétisées.

Cette table pourra ensuite être fusionnée avec `application_train` grâce à la clé :


SK_ID_CURR

# Traitement des valeurs manquantes

Une analyse des valeurs manquantes a été réalisée afin d'identifier les variables nécessitant un traitement spécifique.

Les valeurs manquantes ont été traitées selon la signification métier des variables.

## Variables financières

Les variables suivantes ont été remplacées par 0 :

| Variable | Pourcentage de valeurs manquantes | Méthode | Justification |
|----------|-----------------------------------|---------|---------------|
| AMT_ANNUITY | 71.47% | Remplacement par 0 | Une absence de valeur peut correspondre à une absence d'annuité enregistrée |
| AMT_CREDIT_MAX_OVERDUE | 65.51% | Remplacement par 0 | Absence de montant de retard maximal enregistré |
| AMT_CREDIT_SUM_LIMIT | 34.47% | Remplacement par 0 | Absence de limite de crédit disponible |
| AMT_CREDIT_SUM_DEBT | 15.01% | Remplacement par 0 | Absence de dette connue |

Des variables indicatrices de valeurs manquantes ont également été créées afin de conserver l'information liée à l'absence de donnée.

Exemple :

| AMT_CREDIT_SUM_DEBT | AMT_CREDIT_SUM_DEBT_MISSING |
|--------------------|-----------------------------|
| 50000 | 0 |
| 0 | 1 |

---

## Variables temporelles

Les variables temporelles suivantes :

- `DAYS_CREDIT_ENDDATE`
- `DAYS_ENDDATE_FACT`

ont été complétées par la médiane.

La médiane a été choisie car elle est moins sensible aux valeurs extrêmes et permet de conserver une distribution plus stable.

---

## Vérification après traitement

Après l'imputation :

- aucune valeur manquante restante dans la table `bureau` ;
- les informations liées aux absences ont été conservées grâce aux indicateurs `_MISSING`.

La table obtenue est donc prête pour la création des variables agrégées par client.
